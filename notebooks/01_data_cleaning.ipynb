### Converting Response Time Values into Seconds

After standardising the text, valid response-time observations were converted into a single numerical unit: seconds. Using one consistent unit avoids ambiguity and makes the variable suitable for statistical analysis and machine-learning models.

For approximate ranges, the midpoint was used. For example, `3–4 minutes` was represented as 210 seconds. Textual non-responses were treated as missing values because they did not provide a measurable response time.

```python
def convert_response_time_to_seconds(value):
    if pd.isna(value):
        return np.nan

    value = str(value).lower().strip()

    # Correct common spelling variations
    value = (
        value
        .replace("mintues", "minutes")
        .replace("minitues", "minutes")
    )

    non_response_values = {
        "did not answer",
        "he did not response",
        "just said hello",
        "not response",
        "nothing"
    }

    if value in non_response_values:
        return np.nan

    # Approximate ranges: use midpoint
    if value == "3-4  minutes":
        return 3.5 * 60

    if value == "4-5 minutes":
        return 4.5 * 60

    if value == "about 10 to 20 seconds":
        return 15

    minute_match = re.search(r"(\d+(?:\.\d+)?)\s*minute", value)
    minutes = float(minute_match.group(1)) if minute_match else 0

    second_match = re.search(r"(\d+(?:\.\d+)?)\s*(?:second|sec)", value)
    seconds = float(second_match.group(1)) if second_match else 0

    total_seconds = (minutes * 60) + seconds

    return total_seconds if total_seconds > 0 else np.nan
```
